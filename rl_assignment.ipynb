# Custom Reinforcement Learning Agent Implementation using Farama Gymnasium

This notebook trains a Q-learning agent on the `FrozenLake-v1` environment and visualizes the learning curve and a short policy rollout.

```python
import os
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import gymnasium as gym
from agent import QLearningAgent
from utils import evaluate_policy, plot_training_curve, save_policy_gif

np.random.seed(1)
OUTPUT_DIR = os.getcwd()
ENV_NAME = 'FrozenLake-v1'
```

```python
env = gym.make(ENV_NAME, is_slippery=False, render_mode='rgb_array')
agent = QLearningAgent(
    n_states=env.observation_space.n,
    n_actions=env.action_space.n,
    learning_rate=0.8,
    discount_factor=0.99,
    epsilon=1.0,
    epsilon_min=0.05,
    epsilon_decay=0.995,
)
rewards = agent.train(env, episodes=3000, max_steps=200)
print('Episodes trained:', len(rewards))
print('Average reward in last 100 episodes:', sum(rewards[-100:]) / 100)
```

```python
plot_training_curve(rewards, os.path.join(OUTPUT_DIR, 'training_progress.png'))
plt.figure(figsize=(10, 5))
plt.plot(rewards, alpha=0.3, label='Episode reward')
plt.title('FrozenLake Training Curve')
plt.xlabel('Episode')
plt.ylabel('Reward')
plt.grid(alpha=0.2)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
avg_return, success_rate = evaluate_policy(env, agent.q_table, episodes=200)
print(f'Average return over 200 evaluation episodes: {avg_return:.3f}')
print(f'Success rate: {success_rate:.3f}')
```

```python
save_policy_gif(env, agent.q_table, os.path.join(OUTPUT_DIR, 'trained_agent.gif'), max_steps=200)
print('Saved GIF to', os.path.join(OUTPUT_DIR, 'trained_agent.gif'))
agent.save(os.path.join(OUTPUT_DIR, 'frozenlake_q_table.npy'))
print('Saved model weights to', os.path.join(OUTPUT_DIR, 'frozenlake_q_table.npy'))
```

```python
env.close()
print('Notebook completed successfully.')
```
